```python
Balance = 50000
PIN = 1234


class Account:

    def __init__(self, pin):
        self.pin = pin
        self.balance = Balance


class CurrentAccount(Account):

    def deposit(self):

        amount = int(input("How much do you want to deposit? "))

        self.balance += amount

        print("Your new balance is", self.balance)


    def withdraw(self):

        amount = int(input("How much do you want to withdraw? "))

        limit = 5000

        if amount > limit:
            print("You have exceeded the withdrawal limit.")

        elif amount <= self.balance:
            self.balance -= amount
            print("Your new balance is", self.balance)

        else:
            print("Insufficient funds.")


class SavingsAccount(Account):

    def deposit(self):

        amount = int(input("How much do you want to deposit? "))

        self.balance += amount

        print("Your new balance is", self.balance)


    def withdraw(self):

        amount = int(input("How much do you want to withdraw? "))

        limit = 15000

        if amount > limit:
            print("You have exceeded the withdrawal limit.")

        elif amount <= self.balance:
            self.balance -= amount
            print("Your new balance is", self.balance)

        else:
            print("Insufficient funds.")


pin = int(input("Please enter your PIN: "))

if pin != PIN:

    print("Incorrect PIN. Process stopped.")

else:


    account_type = input("Do you have a current account or a savings account? ")


    if account_type == "current" or account_type == "Current" or account_type == "CURRENT":

        account = CurrentAccount(pin)
        print("Welcome to your current account!")

    elif account_type == "savings" or account_type == "Savings" or account_type == "SAVINGS":

        account = SavingsAccount(pin)
        print("Welcome to your savings account!")

    else:

        print("Invalid account type. Process stopped.")


c1 = CurrentAccount(PIN)
c1.deposit()
```
